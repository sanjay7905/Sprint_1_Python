Logging is the process of recording events, messages, warnings, and errors that occur while a program runs.

<h2>Why is Logging used?</h2>
    
Logging helps us:

track program flow

<h1>Log Levels</h1>

<h2>DEBUG</h2>

Detailed information for debugging.

<h2>INFO</h2>

Normal program events.

<h2>WARNING</h2>

Something unexpected happened, but the program can continue.

<h2>ERROR</h2>
    
A serious problem happened in one operation.

<h2>CRITICAL</h2>
    
A very serious problem that may stop the program.

<h2>Formatter</h2>
    
A formatter decides how the log message should look.

Example: 2026-07-10 15:10:00 | INFO | Application started

A formatter makes logs:

readable
consistent
useful for debugging

```python
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
)

logging.info("Application started")
logging.warning("This is a warning")
logging.error("This is an error")
```

<h2>Handlers</h2>

A handler decides where the log message goes.

<h2>Console Handler</h2>
    
Sends logs to the screen or notebook output.

<h2>File Handler</h2>
    
Sends logs to a file so they can be saved and checked later.

```python
logger = logging.getLogger("logging_example")
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter("%(asctime)s | %(name)s | %(levelname)s | %(message)s")

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.DEBUG)
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler("logs/application.log")
file_handler.setLevel(logging.INFO)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Checking values")
logger.info("Task completed")
logger.warning("File is empty")
logger.error("File not found")
```

<h2>Rotating Logs</h2>

Rotating logs automatically create backup files when a log file becomes too large.

<h2>Why is this useful?</h2>
    
It prevents:

huge log files
difficult file management
storage problems

```python
import logging
from logging.handlers import RotatingFileHandler
import os

rot_logger = logging.getLogger("rotation_demo")
rot_logger.setLevel(logging.DEBUG)

for handler in rot_logger.handlers[:]:
    handler.close()
    rot_logger.removeHandler(handler)

rot_handler = RotatingFileHandler(
    "logs/rotating.log",
    maxBytes=200,
    backupCount=3,
)

rot_handler.setFormatter(formatter)
rot_handler.setLevel(logging.DEBUG)

rot_logger.addHandler(rot_handler)

for i in range(1, 25):
    rot_logger.info(f"Log message {i}")
```

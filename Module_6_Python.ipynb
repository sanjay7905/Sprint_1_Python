<h2>Constraints</h2>
Constraints are rules applied to table columns to ensure the data is accurate and valid.Constraints are applied while creating the table and inserting the values.

Common Constraints:

NOT NULL – The column cannot have an empty value.

UNIQUE – All values in the column must be different.

PRIMARY KEY – Uniquely identifies each row.

FOREIGN KEY – Connects one table to another.

CHECK – Ensures values meet a condition.

DEFAULT – Assigns a default value if no value is provided.

Example:

CREATE TABLE Students ( Student_ID INT PRIMARY KEY, Name VARCHAR(50) NOT NULL, Age INT CHECK (Age >= 18), City VARCHAR(30) DEFAULT 'Chennai' );

In this example,

PRIMARY KEY makes Student_ID unique.

NOT NULL ensures Name cannot be empty.

CHECK ensures Age is at least 18.

DEFAULT sets City to "Chennai" if no value is given

<h2>Queries</h2>

<h3>SELECT</h3>

SELECT is used to retrieve data from a table.

Example:

SELECT Name, Age

FROM Students;

This displays the Name and Age columns from the Students table.

<h2>WHERE</h2>

WHERE is used to filter rows based on a condition.

Example:

SELECT *

FROM Students

WHERE Age > 20;

This displays only students whose age is greater than 20.

<h2>ORDER BY</h2>

ORDER BY is used to sort the result in ascending (ASC) or descending (DESC) order.

Example:

SELECT *

FROM Students

ORDER BY Age ASC;

This sorts students by age from smallest to largest.

<h2>GROUP BY</h2>

GROUP BY is used to group rows that have the same values in a column.

Example:

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department;

This shows the number of employees in each department.

<h2>HAVING</h2>

HAVING is used to filter grouped data after using GROUP BY.

Example:

SELECT Department, COUNT(*)

FROM Employees

GROUP BY Department

HAVING COUNT(*) > 5;

This displays only departments that have more than 5 employees.

<h2>DISTINCT</h2>

DISTINCT is used to return only unique values by removing duplicates.

Example:

SELECT DISTINCT City

FROM Students;

If multiple students are from Chennai, it displays Chennai only once.

<h2>LIMIT</h2>

LIMIT is used to restrict the number of rows returned.

Example:

SELECT *

FROM Students

LIMIT 5;

This displays only the first 5 rows from the Students table.

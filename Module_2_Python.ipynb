<h1>Class</h1>

<h2>Understand the Concept</h2>

<h2>What is a Class?</h2>

A Class is a blueprint or template used to create objects in Python. It defines the attributes (data) and methods (behavior) that objects created from it will have. Instead of writing the same code repeatedly for similar entities, a class allows us to define the structure once and create multiple objects whenever needed.

Classes are one of the fundamental concepts of Object-Oriented Programming (OOP). They help organize code by grouping related data and functionality together, making programs more modular, reusable, and easier to maintain.

In real-world software development, classes are used to represent entities such as students, employees, products, bank accounts, and machine learning models. Each object created from a class shares the same structure but can store different values.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider the blueprint of a house. A blueprint contains the design, structure, and specifications required to build a house, but it is not the actual house itself. Using the same blueprint, multiple houses can be constructed with different owners, colors, or interiors.

Similarly, a Class acts as the blueprint, while the Objects are the actual houses created from it. Every object follows the structure defined by the class but contains its own unique data.

This analogy demonstrates that a class provides a common structure for creating multiple independent objects.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class and creates one or more objects using that class. Each object stores its own data while sharing the same attributes and methods defined in the class.

<h2>Which part of the code demonstrates the concept?</h2>

The class keyword demonstrates the concept of a class by defining the blueprint. The objects created from the class show how multiple instances can be generated using the same structure.

<h2>How does the concept work in this implementation?</h2>

When the program executes, Python first creates the class definition. After that, objects are created from the class. Each object stores its own values and can access the methods defined inside the class. Although all objects share the same structure and behavior, they maintain independent data.

This implementation demonstrates how a class acts as a reusable blueprint for creating multiple objects in Object-Oriented Programming.

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        if self.marks >= 35:
            print(self.name, "has Passed")
        else:
            print(self.name, "has Failed")


s1 = Student("Aishu", 28)

s1.result()
```

<h1>Object</h1>

<h2>Understand the Concept</h2>

<h2>What is an Object?</h2>

An Object is an instance of a class. It is the actual entity created using the blueprint defined by a class. When an object is created, memory is allocated for it, and it can store its own data while using the methods provided by the class.

Objects allow us to represent real-world entities in a program. Although multiple objects are created from the same class, each object maintains its own independent state and values. This enables developers to model multiple entities with the same structure but different data.

In Python, objects are created by calling the class, and they can access the attributes and methods defined within that class.



<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Imagine a car showroom where the company manufactures cars using the same design. The design or blueprint represents the Class, while each individual car produced is an Object.

Every car has the same basic components such as an engine, wheels, and steering, but each car has its own registration number, color, owner, and mileage.

Similarly, every object follows the structure defined by the class while maintaining its own unique data.

This analogy demonstrates that objects are real instances created from a common blueprint.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates one or more objects from a class. Each object stores its own values and can access the methods defined within the class.

<h2>Which part of the code demonstrates the concept?</h2>

The statements used to create objects from the class demonstrate the concept of an object. Each object is an independent instance with its own data while sharing the same functionality provided by the class.

<h2>How does the concept work in this implementation?</h2>

When the class is called, Python creates a new object in memory. Each object stores its own attribute values and can execute the methods defined in the class. Although the objects are created from the same class, they operate independently without affecting each other's data.

This implementation demonstrates how objects are created from a class and how they represent individual entities in Object-Oriented Programming.

```python
class Number:

    def __init__(self, num):
        self.num = num

    def check(self):
        if self.num % 2 == 0:
            print(self.num, "is Even")
        else:
            print(self.num, "is Odd")


n1 = Number(25)

n1.check()
```

<h1>Constructor (__init__)</h2>

<h2>Understand the Concept</h2>

<h2>What is a Constructor (init)?</h2>

A Constructor is a special method in Python that is automatically executed when an object is created. It is defined using the __init__() method and is mainly used to initialize the attributes of an object with their initial values.

The constructor eliminates the need to assign values separately after creating an object. Instead, the required information can be provided during object creation, making the code more organized and efficient.

In Object-Oriented Programming, constructors ensure that every object starts with a valid and consistent state. Since the constructor is called automatically whenever an object is instantiated, it simplifies object initialization and improves code readability.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider purchasing a new mobile phone. When you switch it on for the first time, you are asked to configure settings such as language, region, Wi-Fi, and user account. Once these initial settings are completed, the phone is ready to use.

Similarly, when an object is created, the constructor automatically initializes the object's data so that it is ready for use without requiring additional setup.

This analogy demonstrates that the constructor performs the initial configuration of an object immediately after it is created.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class with a constructor and creates one or more objects by passing values during object creation. The constructor automatically initializes the object's attributes using the provided values.

<h2>Which part of the code demonstrates the concept?</h2>

The __init__() method demonstrates the concept of a constructor. The parameters passed while creating an object are received by the constructor and assigned to the object's attributes.

<h2>How does the concept work in this implementation?</h2>

When an object is created, Python automatically invokes the __init__() method. The constructor receives the values passed during object creation and initializes the object's attributes. Once initialization is complete, the object is fully prepared to use its methods and data.

This implementation demonstrates how constructors automate object initialization and ensure that every object starts with the required information.

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def vote(self):
        if self.age >= 18:
            print(self.name, "is Eligible to Vote")
        else:
            print(self.name, "is Not Eligible to Vote")


p1 = Person("Rahul", 17)
p1.vote()
```

<h1>Instance Variables</h1>

<h2>Understand the Concept</h2>

<h2>What are Instance Variables?</h2>
Instance Variables are variables that belong to an individual object of a class. They are usually initialized inside the __init__() constructor using the self keyword. Each object created from the same class has its own separate copy of these variables.

Instance variables are used to store data that is unique to each object. Since every object maintains its own values, modifying the instance variable of one object does not affect the instance variables of other objects.

In Python, instance variables make it possible to represent different real-world entities using the same class while allowing each object to maintain its own state and information.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a classroom where every student has a name, roll number, and marks. Although all students belong to the same class, each student has different information.

The Student class represents the common structure, while each student's name, roll number, and marks are unique to that individual student.

Similarly, instance variables store information that is specific to each object, allowing every object to maintain its own independent data.

This analogy demonstrates that instance variables belong to individual objects rather than being shared among all objects.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class with instance variables and creates multiple objects. Each object stores its own values for the instance variables, demonstrating that every object maintains its own independent data.

<h2>Which part of the code demonstrates the concept?</h2>

The variables initialized using self inside the __init__() method demonstrate the concept of instance variables. Each object receives its own copy of these variables during object creation.

<h2>How does the concept work in this implementation?</h2>

When an object is created, the constructor initializes the instance variables using the values passed as arguments. Since each object has its own memory allocation, changing the instance variables of one object does not affect the data stored in other objects.

This implementation demonstrates how instance variables store object-specific information and help represent multiple independent entities using the same class.

```python
class Rectangle:

    def __init__(self, length, width):
        self.length = length    # Instance Variable
        self.width = width      # Instance Variable

    def area(self):
        print("Area =", self.length * self.width)


r1 = Rectangle(10, 5)
r1.area()
```

<h1>Class Variables</h1>

<h2>Understand the Concept</h2>

<h2>What are Class Variables?</h2>

Class Variables are variables that belong to the class itself rather than to individual objects. They are declared directly inside the class but outside any methods. Since they are associated with the class, all objects created from that class share the same variable.

Class variables are used to store data that should be common across every object. Examples include company names, tax rates, school names, or the total number of objects created. If the value of a class variable is changed, the updated value becomes available to all objects unless an object overrides it with its own instance variable.

Using class variables helps avoid duplication of shared data and ensures consistency across all objects created from the same class.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a company where every employee works for the same organization named ABC Technologies. While each employee has a unique name, employee ID, and salary, the company name remains the same for everyone.

Instead of storing the company name separately for every employee, it can be stored once and shared by all employees.

Similarly, a class variable stores information that is common to every object created from the class.

This analogy demonstrates that class variables represent shared data, while individual object details are stored separately.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class variable and creates multiple objects. Each object accesses the same shared variable while maintaining its own individual data through instance variables.

<h2>Which part of the code demonstrates the concept?</h2>

The variable declared directly inside the class and outside all methods demonstrates the concept of a class variable. All objects access this shared variable through the class or through their respective objects.

<h2>How does the concept work in this implementation?</h2>

When the class is created, Python allocates memory for the class variable only once. Every object created from the class refers to this shared variable. If the class variable is modified using the class name, the updated value becomes visible to all objects that access it.

This implementation demonstrates how class variables store shared information efficiently and provide a common value across all objects of a class.

```python
class Employee:

    company = "Infosys"      # Class Variable

    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name, "works at", Employee.company)


e1 = Employee("Arjun")
e2 = Employee("sai")

e1.display()
e2.display()
```

<h1>Instance Methods</h1>

<h2>Understand the Concept</h2>

<h2>What are Instance Methods?</h2>

Instance Methods are methods that operate on individual objects of a class. They are defined with the self parameter, which refers to the current object that calls the method. Through self, instance methods can access and modify the instance variables of that specific object.

Instance methods are used to perform operations that depend on an object's unique data. Since each object has its own instance variables, the same instance method can produce different results depending on which object invokes it.

In Python, instance methods are the most commonly used type of methods because they allow objects to interact with and manage their own data.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a bank where every customer has a separate bank account. Although all customers can perform operations such as depositing or withdrawing money, each operation affects only that customer's account balance.

The deposit and withdrawal operations represent instance methods, while the account balance represents instance variables.

This analogy demonstrates that instance methods work on the data of the specific object that calls them, without affecting other objects.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines one or more instance methods within a class and creates multiple objects. Each object calls the same method, but the result depends on the values stored in that particular object's instance variables.

<h2>Which part of the code demonstrates the concept?</h2>

The methods defined with the self parameter demonstrate the concept of instance methods. The use of self to access or modify instance variables shows that the method operates on the current object.

<h2>How does the concept work in this implementation?</h2>

When an object calls an instance method, Python automatically passes that object as the self argument. The method then accesses or updates the instance variables associated with that object. Since every object has its own data, the same method behaves independently for each object.

This implementation demonstrates how instance methods enable objects to perform actions based on their own data while sharing the same method definition.

```python
class Bank:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Updated Balance =", self.balance)


b1 = Bank(5000)
b1.deposit(1000)
```

<h2>Static Methods</h2>

<h2>Understand the Concept</h2>

<h2>What are Static Methods?</h2>

Static Methods are methods that belong to a class but do not operate on instance variables or class variables. They are defined using the @staticmethod decorator and do not require the self or cls parameters.

Static methods are used when a function is logically related to a class but does not need to access or modify the object's data or the class's shared data. They are commonly used for utility functions, helper methods, calculations, validations, and data conversions.

Using static methods improves code organization by keeping related functionality inside the class while avoiding unnecessary dependencies on object or class data.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a calculator application. A function that adds two numbers does not need to know who is using the calculator or any information about the calculator itself. It simply performs the calculation and returns the result.

Similarly, a static method performs a specific task without depending on the object's data or the class's shared data.

This analogy demonstrates that static methods provide utility functionality that is associated with the class but independent of its objects.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a static method inside a class to perform a specific task that does not depend on object-specific or class-specific data. The method is called directly using the class name without creating an object.

<h2>Which part of the code demonstrates the concept?</h2>

The @staticmethod decorator demonstrates the concept of a static method. The absence of the self and cls parameters shows that the method does not interact with instance variables or class variables.

<h2>How does the concept work in this implementation?</h2>

When the static method is called, Python executes it like a regular function associated with the class. Since it does not receive the current object (self) or the class (cls), it cannot access instance variables or class variables directly. Instead, it performs the required operation independently and returns the result.

This implementation demonstrates how static methods are used to organize utility functions within a class while keeping them independent of object and class data.

```python
class Number:

    @staticmethod
    def check(num):
        if num % 2 == 0:
            print("Even Number")
        else:
            print("Odd Number")


Number.check(12)
```

<h1>Class Methods</h1>

<h2>Understand the Concept</h2>

<h2>What are Class Methods?</h2>

Class Methods are methods that operate on the class itself rather than on individual objects. They are defined using the @classmethod decorator and receive the class as their first parameter, conventionally named cls.

Class methods are used when a method needs to access or modify class variables, create objects in alternative ways, or perform operations that apply to the entire class. Since they work with the class instead of a specific object, they can be called using either the class name or an object.

In Python, class methods are commonly used as factory methods to create objects with predefined configurations and to manage data shared by all instances.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a university where every student belongs to the same institution. If the university changes its official name, the change applies to all students rather than to a single student.

Instead of updating the university name for every student individually, the change is made once at the university level.

Similarly, a class method works with data that belongs to the entire class rather than data stored in individual objects.

This analogy demonstrates that class methods are used to manage information shared across all objects of a class.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class method that accesses or modifies a class variable. It demonstrates how operations can be performed at the class level without depending on the data of individual objects.

<h2>Which part of the code demonstrates the concept?</h2>

The @classmethod decorator and the cls parameter demonstrate the concept of a class method. The use of cls to access or modify class variables shows that the method operates on the class itself.

<h2>How does the concept work in this implementation?</h2>

When the class method is called, Python automatically passes the class as the cls argument. The method then accesses or updates the class variables associated with the class. Since these variables are shared among all objects, any changes made through the class method are reflected across every instance of the class.

This implementation demonstrates how class methods manage shared data and perform class-level operations efficiently.

```python
class Student:

    school = "ABC School"

    @classmethod
    def change_school(cl):
        cl.school = "XYZ School"


print(Student.school)

Student.change_school()
```

<h1>Properties (@property)</h1>

<h2>Understand the Concept</h2>

<h2>What are Properties (@property)?</h2>

A Property in Python is a feature that allows a method to be accessed like an attribute using the @property decorator. It provides a controlled way to read, modify, or validate an object's data without exposing the internal implementation.

Properties are mainly used to implement encapsulation. Instead of allowing direct access to instance variables, properties enable developers to define getter, setter, and deleter methods. This helps enforce validation rules, maintain data integrity, and prevent invalid values from being assigned to an object.

In Python, properties make code cleaner and more maintainable because users of the class can access values as attributes while the class internally controls how those values are managed.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider an online banking application where a customer can view or update personal information. Before saving a new mobile number or email address, the system validates the input to ensure it follows the required format. If the input is invalid, the update is rejected.

Similarly, a property allows controlled access to an object's data by validating or processing values before they are stored or retrieved.

This analogy demonstrates that properties protect an object's data while providing a simple and user-friendly interface.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program uses the @property decorator to provide controlled access to an object's attribute. It allows the value to be read, updated, or validated without exposing the internal implementation directly.

<h2>Which part of the code demonstrates the concept?</h2>

The @property decorator defines the getter method, while the @<property_name>.setter decorator defines the setter method. These decorators demonstrate how properties manage access to an object's data.

<h2>How does the concept work in this implementation?</h2>

When the property is accessed, Python automatically calls the getter method instead of directly returning the instance variable. When a new value is assigned, Python invokes the setter method, where validation or additional processing can be performed before updating the data.

This implementation demonstrates how properties provide controlled access to object attributes, improve data security, and support the principles of Object-Oriented Programming.

```python
class Employee:

    def __init__(self, salary):
        self.salary = salary

    @property
    def total_salary(self):
        return self.salary + 5000


e1 = Employee(40000)

print(e1.total_salary)
```

<h1>Encapsulation</h1>

<h2>Understand the Concept</h2>

<h2>What is Encapsulation?</h2>

Encapsulation is one of the fundamental principles of Object-Oriented Programming (OOP). It is the process of combining data (attributes) and the methods (functions) that operate on that data into a single unit called a class. Encapsulation also controls direct access to an object's internal data by restricting how it can be viewed or modified.

In Python, encapsulation is achieved using access modifiers such as public, protected (single underscore _), and private (double underscore __) attributes. Instead of allowing external code to modify an object's data directly, methods are provided to safely access or update the data.

Encapsulation improves data security, maintains data integrity, reduces unintended modifications, and makes code more maintainable. It also allows the internal implementation of a class to change without affecting the code that uses the class.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider an ATM machine. A user can withdraw money, deposit money, or check their account balance using the options provided on the screen. However, the user cannot directly access or modify the bank's database.

The ATM acts as a controlled interface between the user and the bank's internal system. Only authorized operations are allowed, ensuring that sensitive data remains protected.

Similarly, encapsulation hides an object's internal data and allows interaction only through controlled methods, preventing unauthorized or accidental modifications.

This analogy demonstrates how encapsulation protects data while providing a safe and controlled way to interact with an object.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a class that stores its internal data using encapsulation. Instead of accessing the data directly, the program uses methods or properties to retrieve or update the values in a controlled manner.

<h2>Which part of the code demonstrates the concept?</h2>

The private or protected attributes, along with the public methods used to access or modify them, demonstrate the concept of encapsulation. These methods act as a controlled interface between the user and the object's internal data.

<h2>How does the concept work in this implementation?</h2>

When the program creates an object, its internal data is stored within the class. External code cannot directly modify the protected or private attributes. Instead, it must use the provided methods or properties, where validation and additional business logic can be applied before updating the data.

This implementation demonstrates how encapsulation protects an object's internal state, improves data security, and ensures that object data is managed in a safe and controlled manner.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance      # Private Variable

    def deposit(self, amount):
        self.__balance += amount

    def show_balance(self):
        print("Balance =", self.__balance)


b1 = BankAccount(5000)

b1.deposit(1000)
b1.show_balance()
```

<h1>Abstraction</h1>

<h2>Understand the Concept</h2>

<h2>What is Abstraction?</h2>

Abstraction is one of the core principles of Object-Oriented Programming (OOP). It is the process of hiding the internal implementation details of a class while exposing only the essential features required by the user. This allows users to interact with an object without needing to understand how the underlying functionality is implemented.

In Python, abstraction is commonly implemented using the abc (Abstract Base Class) module. Abstract classes define one or more abstract methods that must be implemented by their child classes. An abstract class cannot be instantiated directly, ensuring that subclasses provide their own specific implementation.

Abstraction simplifies complex systems, improves code organization, promotes consistency across related classes, and makes applications easier to maintain and extend.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider driving a car. The driver interacts with the steering wheel, accelerator, brake pedal, and gear lever to operate the vehicle. However, the driver does not need to know the internal working of the engine, transmission, or braking system.

The car exposes only the necessary controls while hiding its internal complexity.

Similarly, abstraction allows users to access only the required functionality of an object while keeping the implementation details hidden.

This analogy demonstrates how abstraction simplifies interaction with complex systems by exposing only what is necessary.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines an abstract class containing one or more abstract methods and creates a subclass that provides concrete implementations for those methods. The subclass objects can then use the implemented functionality.

<h2>Which part of the code demonstrates the concept?</h2>

The abstract class, the @abstractmethod decorator, and the subclass implementing the abstract methods demonstrate the concept of abstraction. The abstract class defines the required behavior, while the child class provides the actual implementation.

<h2>How does the concept work in this implementation?</h2>

When the program runs, the abstract class acts as a template that specifies what methods every subclass must implement. Python prevents direct creation of objects from the abstract class. Instead, objects are created from subclasses that provide complete implementations of all abstract methods. This ensures consistency while allowing different subclasses to implement the same functionality in their own way.

This implementation demonstrates how abstraction hides implementation details, enforces a common interface, and promotes flexibility and maintainability in Object-Oriented Programming.

```python
from abc import ABC, abstractmethod

class Bank(ABC):

    @abstractmethod
    def deposit(self):
        pass


class SBI(Bank):

    def deposit(self):
        print("Amount Deposited Successfully")


s = SBI()
s.deposit()
```

<h1>Inheritance</h1>

<h2>Understand the Concept</h2>

<h2>What is Inheritance?</h2>

Inheritance is one of the fundamental concepts of Object-Oriented Programming (OOP) that allows a new class to acquire the attributes and methods of an existing class. The existing class is called the Parent Class (Base Class), and the new class is called the Child Class (Derived Class).

Inheritance promotes code reusability by allowing common functionality to be defined once in the parent class and reused by multiple child classes. The child class can use the inherited features as they are, extend them by adding new functionality, or modify them by overriding existing methods.

In Python, inheritance helps developers create a logical relationship between classes, reduce code duplication, improve maintainability, and build scalable applications.



<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a family where children inherit certain characteristics from their parents, such as physical features, surname, or family traditions. Although each child inherits common characteristics, they also have their own unique personalities, skills, and interests.

Similarly, in Python, a child class automatically inherits the attributes and methods of the parent class while also having the ability to define its own additional features or customize inherited behavior.

This analogy demonstrates how inheritance enables code reuse while allowing specialization in child classes.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a parent class containing common attributes and methods, and then creates a child class that inherits those features. The child class can access the inherited members and may also include additional methods or override existing ones.

<h2>Which part of the code demonstrates the concept?</h2>

The child class declaration using the parent class name demonstrates inheritance. The inherited methods and attributes accessed by the child object show that the child class automatically receives the functionality of the parent class.

<h2>How does the concept work in this implementation?</h2>

When the child class is created, it automatically inherits the accessible attributes and methods from the parent class. An object of the child class can directly use these inherited members without redefining them. If needed, the child class can also introduce new methods or modify inherited methods to meet specific requirements.

This implementation demonstrates how inheritance promotes code reusability, establishes relationships between classes, and supports the development of modular and maintainable applications.

```python
class Vehicle:

    def start(self):
        print("Vehicle Started")


class Car(Vehicle):
    pass


c = Car()
c.start()
```

<h1>Multiple Inheritance</h1>

<h2>Understand the Concept</h2>

<h2>What is Multiple Inheritance?</h2>

Multiple Inheritance is a type of inheritance in which a single child class inherits attributes and methods from two or more parent classes. This allows the child class to combine the functionalities of multiple classes into one, enabling greater code reuse and flexibility.

In Python, multiple inheritance is supported by allowing multiple parent classes to be specified within parentheses when defining the child class. When two or more parent classes contain methods with the same name, Python follows the Method Resolution Order (MRO) to determine which method should be executed.

Multiple inheritance is useful when a class needs features from different independent classes. However, it should be used carefully because excessive use can make code more complex and harder to maintain.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a smartphone. It functions as a phone for making calls, a camera for taking pictures, and a music player for playing audio. These features originate from different devices but are combined into a single smartphone.

Similarly, in multiple inheritance, a child class combines the functionalities of multiple parent classes into one class, allowing it to perform various tasks without rewriting code.

This analogy demonstrates how multiple inheritance enables a single class to inherit and use capabilities from more than one parent.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines two or more parent classes, each providing different methods or attributes. A child class inherits from all the parent classes, allowing it to access and use the functionality provided by each one.

<h2>Which part of the code demonstrates the concept?</h2>

The child class declaration that specifies multiple parent classes demonstrates multiple inheritance. The child object calling methods inherited from different parent classes confirms that it has successfully inherited functionality from all specified parents.

<h2>How does the concept work in this implementation?</h2>

When the child class is created, Python inherits the accessible members from all the specified parent classes. If methods with the same name exist in multiple parents, Python follows the Method Resolution Order (MRO) to determine which method to execute. The child class can use inherited methods directly or define its own methods to extend or customize functionality.

This implementation demonstrates how multiple inheritance combines features from multiple parent classes, promotes code reuse, and allows a class to perform multiple responsibilities while following Python's method resolution rules.

```python
class Personal:

    def name(self):
        print("Name: Aishu")


class Marks:

    def score(self):
        print("Marks: 90")


class Student(Personal, Marks):
    pass


s = Student()

s.name()
```

<h1>Multilevel Inheritance</h1>

<h2>Understand the Concept</h2>
    
<h2>What is Multilevel Inheritance?</h2>
    
Multilevel Inheritance is a type of inheritance in which a class inherits from another derived class, forming a chain of inheritance. In this hierarchy, one class acts as both a child of its parent class and a parent to another child class.

For example, if Class A is the parent of Class B, and Class B is the parent of Class C, then Class C inherits the features of both Class B and Class A. This creates multiple levels of inheritance where functionality is passed from one generation of classes to the next.

Multilevel inheritance promotes code reuse, reduces duplication, and helps organize classes in a logical hierarchy. It is commonly used when objects share common characteristics while becoming more specialized at each level.



<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider the hierarchy of an educational institution.

A Person has common characteristics such as name and age.
A Teacher is a type of Person and inherits those common characteristics while adding subject-related information.
A Class Teacher is a specialized Teacher who inherits both the Person and Teacher characteristics while adding responsibilities such as managing a classroom.
Here, the inheritance follows the sequence:

Person - Teacher - Class Teacher

This analogy demonstrates how each level inherits the functionality of the previous level while introducing additional features.



<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates a chain of inheritance consisting of a parent class, an intermediate child class, and a final derived class. The final class can access the attributes and methods inherited from every class in the hierarchy.

<h2>Which part of the code demonstrates the concept?</h2>

The class definitions where one child class inherits from another child class demonstrate multilevel inheritance. The object of the final derived class accessing methods from all previous classes confirms that inheritance occurs across multiple levels.

<h2>How does the concept work in this implementation?</h2>

When the final derived class is created, Python first inherits the members of the intermediate class. Since the intermediate class has already inherited the members of the base class, the final derived class automatically gains access to features from both classes. This creates a continuous inheritance chain, allowing the most specialized class to reuse functionality from every level.

This implementation demonstrates how multilevel inheritance builds hierarchical relationships between classes, promotes code reuse, and enables gradual specialization of objects.

```python
class Person:

    def name(self):
        print("Name: Aishu")


class Student(Person):

    def course(self):
        print("Course: Python")


class CollegeStudent(Student):

    def college(self):
        print("College: XYZ Engineering College")


c = CollegeStudent()

c.name()
c.course()
c.college()
```

<h1>Hierarchical Inheritance</h1>

<h2>Understand the Concept</h2>
    
<h2>What is Hierarchical Inheritance?</h2>
    
Hierarchical Inheritance is a type of inheritance in which multiple child classes inherit from a single parent class. The parent class contains common attributes and methods that are shared among all child classes, while each child class can also define its own unique features and behaviors.

This type of inheritance helps eliminate duplicate code because common functionality is written only once in the parent class and reused by all child classes. It also creates a clear hierarchical relationship between classes, making applications easier to organize and maintain.

In Python, hierarchical inheritance is commonly used when different types of objects share the same basic characteristics but also require specialized functionality.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a Person as the parent class. Every person has common attributes such as a name, age, and address. From this parent class, different child classes such as Student, Teacher, and Employee can be created.

Each child class automatically inherits the common details of a person but also includes its own unique information. For example, a Student has a roll number, a Teacher has a subject, and an Employee has a department.

This analogy demonstrates how multiple child classes can share common functionality from a single parent while maintaining their own specialized features.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a parent class with common attributes and methods, then creates multiple child classes that inherit from the same parent. Each child class can use the inherited functionality while also defining additional methods or attributes specific to its purpose.

<h2>Which part of the code demonstrates the concept?</h2>

The multiple child classes inheriting from a single parent class demonstrate hierarchical inheritance. The child objects accessing inherited methods and their own specialized methods show how the inheritance hierarchy works.

<h2>How does the concept work in this implementation?</h2>

When each child class is created, Python automatically inherits the accessible attributes and methods from the parent class. As a result, every child class can use the shared functionality without redefining it. Each child class can then extend the inherited functionality by adding its own attributes or methods to meet specific requirements.

This implementation demonstrates how hierarchical inheritance promotes code reuse, reduces redundancy, and organizes related classes under a common parent while allowing each child class to have specialized behavior.

```python
class Animal:

    def eat(self):
        print("Animal Eats")


class Dog(Animal):

    def sound(self):
        print("Dog Barks")


class Cat(Animal):

    def sound(self):
        print("Cat Meows")


class Cow(Animal):

    def sound(self):
        print("Cow Moos")


class Lion(Animal):

    def sound(self):
        print("Lion Roars")


dog = Dog()
cat = Cat()
cow = Cow()
lion = Lion()

dog.eat()
dog.sound()

cat.eat()
cat.sound()

cow.eat()
cow.sound()

lion.eat()
lion.sound()
```

<h1>Polymorphism</h1>

<h2>Understand the Concept</h2>

<h2>What is Polymorphism?</h2>

Polymorphism is one of the core principles of Object-Oriented Programming (OOP). The word polymorphism means "many forms." It allows the same method, function, or interface to perform different actions depending on the object that uses it.

In Python, polymorphism enables different classes to define methods with the same name while providing their own implementations. When the same method is called on different objects, Python automatically executes the appropriate implementation for each object.

Polymorphism improves flexibility, code reusability, and maintainability by allowing developers to write generic code that works with different types of objects without modifying the existing implementation.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a remote control that can operate different electronic devices such as a television, air conditioner, or music system. Although the Power button is the same, pressing it performs different actions depending on the device being controlled.

Similarly, in polymorphism, the same method name can produce different behaviors depending on the object that invokes it.

This analogy demonstrates how a single interface can represent multiple forms of behavior while keeping the interaction simple and consistent.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines multiple classes that contain methods with the same name but different implementations. When the same method is called on objects of different classes, each object executes its own version of the method.

<h2>Which part of the code demonstrates the concept?</h2>

The methods with the same name implemented in different classes demonstrate polymorphism. Calling the same method on different objects and receiving different outputs confirms that Python selects the appropriate implementation based on the object's type.

<h2>How does the concept work in this implementation?</h2>

When the program executes, Python determines the type of the object that calls the method. It then invokes the corresponding implementation defined within that object's class. Although the method name remains the same, the behavior changes according to the object's class.

This implementation demonstrates how polymorphism allows one interface to represent multiple behaviors, making programs more flexible, reusable, and easier to extend.

```python
class Dog:

    def sound(self):
        print("Dog Barks")


class Cat:

    def sound(self):
        print("Cat Meows")


class Cow:

    def sound(self):
        print("Cow Moos")


class Lion:

    def sound(self):
        print("Lion Roars")


dog = Dog()
cat = Cat()
cow = Cow()
lion = Lion()

dog.sound()
cat.sound()
cow.sound()
lion.sound()
```

<h1>Method Overriding</h1>

<h2>Understand the Concept</h2>

<h2>What is Method Overriding?</h2>

Method Overriding is an Object-Oriented Programming concept in which a child class provides its own implementation of a method that is already defined in its parent class. The child class uses the same method name, the same parameters, and the same return type (where applicable), but changes the behavior according to its own requirements.

Method overriding allows a child class to inherit the common functionality of the parent class while customizing or extending specific behaviors. This helps in achieving runtime polymorphism and improves the flexibility of object-oriented applications.

In Python, when a method is called using an object of the child class, Python automatically executes the overridden method instead of the parent class method.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a Notification system in an application. Every notification sends a message, but the way the message is sent depends on the notification type.

An Email Notification sends the message through email.
An SMS Notification sends the message through text messaging.
A Push Notification sends the message through a mobile application.
Although all notification types perform the same operation—sending a message—each one implements the operation differently.

This analogy demonstrates how the same method can have different implementations in different child classes

<h2>Implementation Explanation<h2>

<h2>What does this program do?</h2>

This program defines a method in the parent class and then redefines the same method in the child class. When the method is called using the child class object, the child's implementation is executed instead of the parent's implementation.

<h2>Which part of the code demonstrates the concept?</h2>

The child class method having the same name as the parent class method demonstrates method overriding. Calling the method through the child object and observing the child-specific output confirms that overriding has taken place.

<h2>How does the concept work in this implementation?</h2>

When the program runs, Python first checks whether the child class contains the requested method. Since the child class has overridden the method, Python executes the child's implementation. If the child class does not override the method, Python automatically uses the inherited version from the parent class.

This implementation demonstrates how method overriding allows child classes to customize inherited behavior while preserving the overall class hierarchy and supporting runtime polymorphism.

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog Barks")


d = Dog()
d.sound()
```

<h1>Method Overloading</h1>

<h2>Understand the Concept</h2>

<h2>What is Method Overloading?</h2>

Method Overloading is the ability to define multiple methods with the same name but different parameter lists. In many programming languages such as Java and C++, method overloading is supported directly by the language.

However, Python does not support traditional method overloading. If multiple methods with the same name are defined in a class, the most recently defined method overrides the previous ones.

To achieve method overloading in Python, developers use alternative approaches such as default arguments, variable-length arguments (*args and **kwargs), or conditional logic inside a single method. These techniques allow one method to handle different numbers or types of arguments while maintaining a single method definition.

This approach provides flexibility and simplifies code by allowing one method to perform multiple related operations.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider an online shopping application where a customer can search for products in different ways.

Search using only the product name.
Search using the product name and category.
Search using the product name, category, and price range.
Although the search operation remains the same, the amount of information provided by the user can vary.

Similarly, Python uses a single method to handle different combinations of input parameters rather than creating multiple methods with the same name.

This analogy demonstrates how Python achieves the behavior of method overloading through flexible parameter handling.



<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a single method that accepts different numbers or types of arguments using default values or variable-length parameters. Based on the provided input, the method performs the appropriate operation.

<h2>Which part of the code demonstrates the concept?</h2>

The use of default parameters, *args, **kwargs, or conditional statements inside a single method demonstrates Python's approach to method overloading. The method adapts its behavior depending on the arguments passed during execution.

<h2>How does the concept work in this implementation?</h2>

When the method is called, Python passes the supplied arguments to the single method definition. The method then checks the number or type of arguments and executes the corresponding logic. This enables one method to handle multiple use cases without defining multiple methods with the same name.

This implementation demonstrates how Python simulates method overloading using flexible parameter handling while maintaining clean, readable, and maintainable code.

```python
class Demo:

    def add(self, a):
        print(a)

    def add(self, a, b):
        print(a + b)


d = Demo()
d.add(10, 20)
```

<h1>Composition</h1>

<h2>Understand the Concept</h2>

<h2>What is Composition?</h2>

Composition is an Object-Oriented Programming (OOP) concept in which one class contains an object of another class as one of its attributes. This represents a strong "has-a" relationship between the two classes. In composition, the contained object is created and managed by the owning class, and its lifecycle depends on the owner.

Composition is used to build complex objects by combining smaller, reusable objects instead of relying solely on inheritance. This approach promotes modularity, code reuse, flexibility, and easier maintenance. If the parent object is destroyed, the composed objects are also destroyed because they exist only as part of the parent object.

In Python, composition is commonly used when one object cannot exist independently without the object that owns it.



<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a House and its Rooms. A house consists of multiple rooms, and the rooms are considered part of that house. If the house is demolished, the rooms also cease to exist because they are an integral part of the house.

Similarly, in composition, one class owns another class, and the contained object exists only as part of the parent object.

This analogy demonstrates the strong "has-a" relationship that defines composition.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates a class that contains an object of another class as one of its attributes. The parent class creates and manages the contained object, demonstrating ownership and a strong relationship between the two classes.

<h2>Which part of the code demonstrates the concept?</h2>

The object of one class being created inside another class demonstrates composition. The parent class is responsible for creating, using, and managing the contained object throughout its lifecycle.

<h2>How does the concept work in this implementation?</h2>

When the parent object is created, it also creates the object that it contains. The contained object performs its assigned responsibilities, but its existence depends entirely on the parent object. If the parent object is removed, the contained object is also removed because it cannot exist independently.

This implementation demonstrates how composition combines multiple classes to build complex systems while maintaining strong ownership, modularity, and code reusability.

```python
class Address:

    def city(self):
        print("City: Chennai")


class Person:

    def __init__(self):
        self.address = Address()    # Composition

    def details(self):
        print("Person Details")
        self.address.city()


p = Person()
p.details()
```

<h1>Aggregation</h1>

<h2>Understand the Concept</h2>

<h2>What is Aggregation?</h2>

Aggregation is an Object-Oriented Programming (OOP) concept in which one class contains a reference to an object of another class, representing a weak "has-a" relationship. Unlike composition, the contained object is created independently and can continue to exist even if the parent object is destroyed.

Aggregation allows multiple classes to share the same object without creating a strong dependency between them. This improves flexibility, promotes code reuse, and makes it easier to design loosely coupled systems.

In Python, aggregation is commonly used when one object uses another object without owning its entire lifecycle.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a Department and its Employees. A department contains multiple employees, but the employees do not depend entirely on the department for their existence. If a department is closed or reorganized, the employees still exist and can be assigned to another department.

Similarly, in aggregation, one class uses another class without owning it completely. Both objects have independent lifecycles.

This analogy demonstrates the weak "has-a" relationship that characterizes aggregation.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates two independent classes. One class receives an object of the other class and uses it to perform its operations. The referenced object is created outside the parent class, demonstrating that both objects exist independently.

<h2>Which part of the code demonstrates the concept?</h2>

Passing an object from one class into another class, typically through the constructor or a method, demonstrates aggregation. The parent class stores a reference to the external object instead of creating it internally.

<h2>How does the concept work in this implementation?</h2>

When the program runs, the independent object is created first. This object is then passed to another class, which uses it to perform specific tasks. Since the referenced object was created independently, it continues to exist even if the parent object is deleted.

This implementation demonstrates how aggregation creates a flexible and loosely coupled relationship between classes while allowing each object to maintain its own independent lifecycle.

```python
class Address:

    def city(self):
        print("City: Chennai")


class Person:

    def __init__(self, address):
        self.address = address

    def display(self):
        print("Person Details")
        self.address.city()


a = Address()
p = Person(a)

p.display()
```

<h1> Association</h1>

<h2>Understand the Concept</h2>

<h2>What is Association?</h2>

Association is an Object-Oriented Programming (OOP) concept that represents a relationship between two or more independent classes. In an association, the objects interact with each other to perform specific tasks, but neither object owns the other. Each object has its own lifecycle and can exist independently.

Association is used to model real-world relationships where objects collaborate without creating a strong dependency. The relationship can be one-to-one, one-to-many, or many-to-many, depending on the application's requirements.

In Python, association helps build flexible and loosely coupled systems by allowing classes to communicate while remaining independent.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a Teacher and a Student. A teacher teaches many students, and a student learns from many teachers. Although they interact with each other, neither the teacher nor the student owns the other. Both can exist independently.

Similarly, in association, different classes work together to achieve a common goal while maintaining their own identities and lifecycles.

This analogy demonstrates how independent objects can collaborate without creating ownership between them.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates two independent classes that interact with each other to complete a task. The objects communicate by passing references, but each object is created and managed independently.

<h2>Which part of the code demonstrates the concept?</h2>
    
The interaction between two independently created objects demonstrates association. One object calls the methods or accesses the functionality of another object without owning or controlling its lifecycle.

<h2>How does the concept work in this implementation?</h2>
    
When the program runs, each object is created separately. One object receives a reference to the other object and uses its methods to perform specific operations. Since both objects are independent, they can continue to exist and function even if the relationship between them changes.

This implementation demonstrates how association enables collaboration between independent classes while maintaining loose coupling, flexibility, and reusability.

```python
class Customer:

    def __init__(self, name):
        self.name = name


class Bank:

    def serve(self, customer):
        print(customer.name, "visited the bank")


c = Customer("Arjun")
b = Bank()

b.serve(c)
```

<h1>Magic (Dunder) Methods</h1>

<h2>Understand the Concept</h2>

<h2>What are Magic (Dunder) Methods?</h2>

Magic Methods, also known as Dunder (Double Underscore) Methods, are special methods in Python whose names begin and end with double underscores, such as __init__, __str__, __repr__, and __len__. These methods are automatically called by Python when specific operations are performed on an object.

Magic methods allow developers to define how objects should behave with built-in Python operations such as object creation, printing, comparison, arithmetic operations, iteration, and length calculation. Instead of calling these methods directly, Python invokes them automatically whenever the corresponding operation is performed.

Using magic methods makes custom classes behave like Python's built-in data types, resulting in cleaner, more intuitive, and more maintainable code.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a smart card used to enter an office building. When the card is scanned at the entrance, the security system automatically verifies the user's identity and grants or denies access. The user simply scans the card without knowing the internal verification process.

Similarly, when operations such as print(), len(), or object creation are performed in Python, the corresponding magic methods are automatically executed behind the scenes without being called explicitly.

This analogy demonstrates how magic methods provide automatic behavior for common operations while hiding the internal implementation.

<h2>Implementation Explanation</h2>
    
<h2>What does this program do?</h2>

This program defines one or more magic methods inside a class to customize how objects behave during standard Python operations such as object creation, printing, or calculating length.

<h2>Which part of the code demonstrates the concept?</h2>

The methods whose names begin and end with double underscores (__) demonstrate the concept of magic methods. Python automatically invokes these methods when the related operation is performed on an object.

<h2>How does the concept work in this implementation?</h2>

When the program performs an operation such as creating an object, printing an object, or calling len() on an object, Python automatically searches for the corresponding magic method inside the class. If the method exists, Python executes it and uses the returned result for that operation.

This implementation demonstrates how magic methods allow developers to customize the behavior of objects while integrating seamlessly with Python's built-in functionality.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


s = Student("Arjun")

print(s)
```

<h1>super()</h1>

<h2>Understand the Concept</h2>

<h2>What is super()?</h2>

super() is a built-in Python function used to access the methods and constructors of a parent class from within a child class. It allows a child class to reuse the functionality of its parent class without directly referring to the parent's class name.

The most common use of super() is to call the parent class constructor (__init__()) so that the parent class attributes are initialized before adding child-specific attributes. It can also be used to call overridden methods from the parent class.

Using super() improves code reusability, reduces duplication, and makes inheritance easier to maintain. It is especially useful in multiple inheritance because it follows Python's Method Resolution Order (MRO) to determine which parent method should be called.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a company where every new employee must first complete the organization's standard onboarding process before joining a specific department. After completing the common onboarding, the employee receives department-specific training.

Similarly, in inheritance, a child class can use super() to execute the parent class initialization first and then perform its own additional initialization.

This analogy demonstrates how super() enables a child class to reuse the parent's functionality before extending it with its own features.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program creates a parent class and a child class. The child class uses the super() function to call the parent class constructor or methods before performing its own operations.

<h2>Which part of the code demonstrates the concept?</h2>

The use of the super() function inside the child class demonstrates the concept. It invokes the parent class constructor or method without directly referencing the parent class name.

<h2>How does the concept work in this implementation?</h2>

When an object of the child class is created, Python executes the child class constructor. Inside the constructor, super() calls the parent class constructor, allowing the parent class to initialize its attributes. After the parent initialization is complete, the child class initializes its own attributes and methods.

This implementation demonstrates how super() promotes code reuse, supports inheritance, reduces duplication, and ensures proper initialization of parent and child classes.

```python
class Animal:

    def show(self):
        print("This is an Animal")


class Dog(Animal):

    def show(self):
        super().show()
        print("This is a Dog")


d = Dog()
d.show()
```

<h1>str()</h1>

<h2>Understand the Concept</h2>

<h2>What is str()?</h2>

The __str__() method is a special (magic or dunder) method in Python that defines the human-readable string representation of an object. It is automatically called whenever the print() function or the str() function is used on an object.

By default, printing an object displays its memory location, which is not meaningful to users. By implementing the __str__() method, developers can provide a clear, readable, and user-friendly description of an object's data.

The __str__() method improves code readability, simplifies debugging, and provides meaningful output when objects are displayed to end users.

<h2>Demonstrate the Concept</h2>

This analogy demonstrates how __str__() improves the presentation of object information for users.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines the __str__() method within a class to return a descriptive string representing the object's data. When the object is printed, Python automatically calls this method and displays the returned string.

<h2>Which part of the code demonstrates the concept?</h2>

The implementation of the __str__() method demonstrates the concept. Calling print(object) or str(object) automatically invokes this method and displays the customized output.

<h2>How does the concept work in this implementation?</h2>

When the program executes the print() function on an object, Python first checks whether the class defines the __str__() method. If it exists, Python calls the method and prints the returned string instead of the default object representation. This allows developers to provide meaningful information about an object in a simple and readable format.

This implementation demonstrates how __str__() enhances object readability by providing a customized, human-friendly representation of an object's data.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name


s = Student("Aishu")

print(s)
```

<h1>repr()</h1>

<h2>Understand the Concept<h2>

<h2>What is repr()?</h2>

The __repr__() method is a special (magic or dunder) method in Python that defines the official string representation of an object. It is automatically called when the repr() function is used or when an object is displayed in the Python interpreter.

Unlike __str__(), which is designed for end users, __repr__() is intended for developers. It should return a detailed and unambiguous representation of an object that is useful for debugging and development. Ideally, the returned string should be capable of recreating the object if passed to Python.

If a class does not define the __str__() method, Python uses the result of __repr__() as the default string representation.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider the packaging label of an electronic device. While a customer may only need to see the product name, the manufacturer requires detailed information such as the model number, serial number, manufacturing date, and hardware specifications for maintenance and troubleshooting.

Similarly, the __repr__() method provides a detailed representation of an object that is primarily intended for developers rather than end users.

This analogy demonstrates how __repr__() supports debugging by providing precise information about an object.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines the __repr__() method within a class to return a detailed string representation of the object's state. When the repr() function is called or the object is inspected in the Python interpreter, this method is automatically executed.

<h2>Which part of the code demonstrates the concept?</h2>

The implementation of the __repr__() method demonstrates the concept. Calling repr(object) or displaying the object in the interactive Python shell invokes this method and returns the developer-oriented representation.

<h2>How does the concept work in this implementation?</h2>

When Python needs the official representation of an object, it first checks whether the class implements the __repr__() method. If available, Python calls the method and uses the returned string to represent the object. This information helps developers inspect object values, understand program behavior, and debug applications more effectively.

This implementation demonstrates how __repr__() provides an accurate and informative representation of an object for development, testing, and debugging purposes.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"


s = Student("Aishu")

print(repr(s))
```

<h1>len()</h1>

<h2>Understand the Concept</h2>

<h2>What is len()?</h2>

The __len__() method is a special (magic or dunder) method in Python that defines how the built-in len() function behaves for objects of a class. By implementing this method, developers can specify what should be considered the "length" of an object.

The __len__() method must return a non-negative integer. Python automatically calls this method whenever the len() function is used on an object. Without this method, calling len() on a custom object raises a TypeError.

This method is commonly used in custom collection classes such as lists, queues, libraries, shopping carts, and inventory systems, where the length represents the number of stored items.

This implementation demonstrates how __len__() integrates custom classes with Python's built-in len() function, making objects more intuitive and consistent to use.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __len__(self):
        return len(self.name)


s = Student("Aishu")

print(len(s))
```

<h1>Dataclasses</h1>

<h2>Understand the Concept</h2>

<h2>What are Dataclasses?</h2>

A Dataclass is a feature introduced in Python 3.7 through the dataclasses module. It is used to simplify the creation of classes that primarily store data. By applying the @dataclass decorator, Python automatically generates common methods such as __init__(), __repr__(), __eq__(), and others, reducing the amount of boilerplate code developers need to write.

Dataclasses improve code readability, reduce duplication, and make programs easier to maintain. They are especially useful when a class is mainly intended to hold data rather than implement complex behavior.

In Python, dataclasses are widely used in AI/ML applications, APIs, configuration management, and data modeling because they provide a clean and efficient way to represent structured data.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider a student registration form. Every student record contains fields such as name, roll number, department, and email. Instead of manually creating a separate process to manage each field, the registration system automatically organizes and stores the information in a structured format.

Similarly, a dataclass automatically creates common methods required to manage data objects, allowing developers to focus on defining the data rather than writing repetitive code.

This analogy demonstrates how dataclasses simplify the creation and management of data-oriented classes.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program defines a dataclass using the @dataclass decorator. The class contains data fields, and Python automatically generates methods such as the constructor, object representation, and comparison methods without requiring explicit implementation.

<h2>Which part of the code demonstrates the concept?</h2>

The @dataclass decorator demonstrates the concept of a dataclass. The class fields defined below the decorator represent the data attributes, while the automatically generated methods are provided by Python.

<h2>How does the concept work in this implementation?</h2>

When the program executes, Python processes the @dataclass decorator before creating the class. Based on the declared fields, it automatically generates methods such as __init__(), __repr__(), and __eq__(). As a result, objects can be created, displayed, and compared without writing these methods manually.

This implementation demonstrates how dataclasses reduce boilerplate code, improve readability, and provide an efficient way to create data-centric classes.

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    salary: int


e = Employee("Rahul", 50000)

print(e)
```

<h1>Singleton Pattern</h1>
<h2>Understand the Concept</h2>

<h2>What is the Singleton Pattern?</h2>

The Singleton Pattern is a software design pattern that ensures a class can have only one object (instance) throughout the lifetime of an application. Whenever an object of the Singleton class is requested, the same instance is returned instead of creating a new one.

The Singleton Pattern is useful when a single shared resource needs to be accessed from different parts of a program. Examples include database connections, application configuration, logging systems, cache managers, and thread pools.

Using the Singleton Pattern helps maintain consistency, reduces unnecessary memory usage, and prevents the creation of multiple objects that could lead to inconsistent application behavior.

<h2>Demonstrate the Concept</h2>

<h2>Real-World Analogy</h2>

Consider the CEO of a company. At any given time, a company has only one CEO. Regardless of how many employees need approval or guidance, they all communicate with the same CEO rather than creating a new CEO for each request.

Similarly, the Singleton Pattern ensures that only one object of a class exists, and every part of the application accesses that same object.

This analogy demonstrates how a single shared instance can coordinate and manage resources consistently across an application.

<h2>Implementation Explanation</h2>

<h2>What does this program do?</h2>

This program implements the Singleton Pattern by ensuring that only one object of the class is created. If an object already exists, the program returns the existing instance instead of creating a new one.

<h2>Which part of the code demonstrates the concept?</h2>

The logic that checks whether an instance already exists and returns the existing object instead of creating a new one demonstrates the Singleton Pattern. This ensures that all variables reference the same object in memory.

<h2>How does the concept work in this implementation?</h2>

When the program attempts to create an object, it first checks whether an instance of the class already exists. If no instance exists, Python creates a new object and stores its reference. If an instance already exists, the existing object is returned. As a result, every request for the class receives the same shared object.

This implementation demonstrates how the Singleton Pattern enforces a single instance, ensures consistent access to shared resources, and improves resource management within an application.

```python
class Logger:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance


log1 = Logger()
log2 = Logger()

print(log1 is log2)
```

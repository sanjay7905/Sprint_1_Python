<h2>while loop</h2>
while loop is used to repeat a block of code as long as a condition is True.

```python
while True:
    print("1. Check Balance")
    print("2. Withdraw")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "3":
        print("Thank you")
        break
```

while loop repeatedly executes a block of code as long as the given condition is true.

<h2>Break</h2>

The break statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

```python
# Example
people = ["Person1", "Person2", "Friend",
          "Person4", "Person5"]

for person in people:
    print("Searching: ", person)
    if person == "Friend":
        print("Found the Friend")
        break
else:
    print("Not found")
```

if I am searching for my friend in a crowd of 50 people and I find my friend at the 10th person, I don't need to search the remaining people.

I use break to stop the loop

<h2>Continue</h2>

The continue statement is used to skip the current iteration of a loop and continue with the next iteration.

```python
# skips negative values

lst = [10,-25,40,-57,89,90,-98]

for number in lst:
    if number<0:
        continue
    print(number)
```

Imagine a teacher checking homework. <br>

One student forgot the homework. <br>

The teacher skips that student and continues checking the rest of the class. <br>

The checking doesn't stop it simply skips one student.

<h2>pass</h2>

 pass statement is a placeholder that does nothing. It is used when Python requires a statement, but no action needs to be performed.

<h4>Real-life example</h4>

```python
people = ["Arun", "Rahul", "Kumar", "jay", "Vijay"]

for person in people:

    if person != "jay":
        pass
    else:
        print("Found my friend!")
```

<h2>Recursive Functions</h2>

A recursive function is a function that calls itself to solve a problem. It continues calling itself until a stopping condition (called the base case) is reached.

Example

Imagine standing between two mirrors.

You see your reflection repeating again and again.

It continues until the reflections become too small to notice.

Similarly, a recursive function keeps calling itself until it reaches the stopping condition.

```python
# factorial of a number
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

<h2>Generators</h2>

A generator is a special type of function that produces values one at a time instead of returning all values at once.

Example

Imagine a ticket machine.

It gives one ticket each time you press the button instead of printing all tickets at once.

A generator works the same way.

```python
def square():
    for i in range(2,6):
        yield i * i

gen = square()

for num in gen:
    print(num)
```

<h2>yield</h2>

The yield keyword is used inside a generator function to return one value at a time while saving the function's state. The function resumes from where it stopped when called again.

Example

Think of a TV series.

You watch one episode today and continue from the next episode tomorrow.

It doesn't start from Episode 1 every time.

Similarly, yield pauses and resumes the function.

```python
def even():
    for i in range(1, 11):
        if i % 2 == 0:
            yield i

for num in even():
    print(num)
```

<h2>Decorators</h2>

A decorator is a function that adds extra functionality to another function without changing its original code.

Example

Think of gift wrapping.

The gift remains the same, but wrapping adds decoration.

Similarly, a decorator adds extra behavior to a function.

```python
def decorator(func):
    def wrapper(a, b):
        print("Adding numbers...")
        func(a, b)
    return wrapper

@decorator
def add(a, b):
    print(a + b)

add(10, 20)
```

<h2>Closures</h2>

A closure is a function that remembers the variables from its outer function even after the outer function has finished executing.

 Example

Imagine your mother gives you a lunch box

Even after you leave home, you still have the lunch box.

Similarly, the inner function remembers data from the outer function.

```python
def multiply(x):
    def inner(y):
        print(x * y)

    return inner

result = multiply(5)
result(4)
```

<h2>global and nonlocal</h2>

global allows a function to modify a variable defined outside all functions (global scope). nonlocal allows an inner function to modify a variable in its enclosing (outer) function.

Example

Imagine a company.

global - Company-wide announcement. nonlocal - Department-level announcement.

A company announcement affects everyone.

A department announcement affects only that department.

```python
balance = 1000

def deposit():
    global balance
    balance += 500

deposit()
print(balance)
```

<h2>any()</h2>

The any() function returns True if at least one element in an iterable is True. Otherwise, it returns False.

Example

```python
numbers = [1, 3, 5, 8]

result = any(num % 2 == 0 for num in numbers)

print(result)

# since 8 is even any() returns true
```

<h2>all()</h2>

The all() function returns True only if every element in an iterable is True. If even one element is False, it returns False.

Example
    
Imagine checking whether all students submitted their assignments.

```python
numbers = [1, 3, 6, 8]

result = all(num % 2 == 0 for num in numbers)

print(result)

# since 1 and 3 are odd all() returns false only if all the numbers in the list satisfies the condition all() returns true
```

<h2>for loop</h2>

for loop is used to repeat a block of code for each item in a sequence or collection.

```python
names = ["kumar", "Arun", "Rahul"]

for name in names:
    print(name)
```

for loop is useful when we need to perform the same task for every item <br>

Example, we can use a for loop to check each student's attendance

<h2>while loop</h2>
while loop is used to repeat a block of code as long as a condition is True.

```python
while True:
    print("1. Check Balance")
    print("2. Withdraw")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "3":
        print("Thank you")
        break
```

while loop repeatedly executes a block of code as long as the given condition is true. 

It stops when the condition becomes false

<h2>Break</h2>

The break statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

```python
# Example
people = ["Person1", "Person2", "Friend",
          "Person4", "Person5"]

for person in people:
    print("Searching: ", person)
    if person == "Friend":
        print("Found the Friend")
        break
else:
    print("Not found")
```

if I am searching for my friend in a crowd of 50 people and I find my friend at the 10th person, I don't need to search the remaining people.

I use break to stop the loop

<h2>Continue</h2>

The continue statement is used to skip the current iteration of a loop and continue with the next iteration.

```python
# skips negative values

lst = [10,-25,40,-57,89,90,-98]

for number in lst:
    if number<0:
        continue
    print(number)
```

Imagine a teacher checking homework. <br>

One student forgot the homework. <br>

The teacher skips that student and continues checking the rest of the class. <br>

The checking doesn't stop it simply skips one student.

<h2>pass</h2>

 pass statement is a placeholder that does nothing. It is used when Python requires a statement, but no action needs to be performed.

<h4>Real-life example</h4>

Imagine you are checking 50 people in a crowd.<br>

If the person is not your friend, do nothing and continue searching

```python
people = ["Arun", "Rahul", "Kumar", "jay", "Vijay"]

for person in people:

    if person != "jay":
        pass
    else:
        print("Found my friend!")
```

<h2>List</h2>

A list is an ordered, mutable collection that can store multiple items. It allows duplicate values and items can be added, removed, or modified.

Ordered - Items keep their position. 
    
Mutable - Items can be changed after creation

```python
#Create a List(Adding names of the student)
student = ["vijay","suriya","rohit","Bhumra"]
print(student)
```

```python
#Add a New Student (append())
student = ["vijay","suriya","rohit","Bhumra"]
student.append("sky")
print(student)
```

```python
#Remove a Student (remove())
student = ["vijay","suriya","rohit","Bhumra"]
student.remove("suriya")
print(student)
```

```python
#Change a name of Student (using Index)
student = ["vijay","suriya","rohit","Bhumra"]
student[2] = "dhoni"
print(student)
```

```python
#Call a particular Student
student = ["vijay","suriya","rohit","Bhumra"]
print(student[1])
```

```python
#Find the Total Number of Students
student = ["vijay","suriya","rohit","Bhumra"]
print(len(student))
```

```python
#Sort in order(Alphabetical order)
student = ["vijay","suriya","rohit","Bhumra"]
student.sort()
print(student)
```

<h2>Tuple</h2>

A tuple is an ordered, immutable collection that stores multiple items. Once created, its values cannot be changed.

Ordered - Items keep their position. 
    
Immutable - Items cannot be modified. 

List ([]) - Can add, remove, and change items. mutable

Tuple (()) - Can only store, access, count, and search items. It cannot be modified after creation. immutable

```python
# Create tuple and access elements
car = ("Tata", "Bugatti","Rolls Royce")
print(car)
print(car[1])
```

```python
# negative indexing
car = ("Tata", "Bugatti","Rolls Royce")
print(car[-1])
```

```python
# slicing
car = ("Tata", "Bugatti","Rolls Royce", "Mustang")
print(car[1:3])
```

```python
# tuple length
car = ("Tata", "Bugatti","Rolls Royce", "Mustang")
print(len(car))
```

```python
# membership
car = ("Tata", "Bugatti","Rolls Royce", "Mustang")
print("Tata" in car)
print("Tata" not in car)
```

```python
#concatenation
car1 = ("Tata", "Bugatti","Rolls Royce", "Mustang")
car2 = ("hyundai", "benz")
print(car1 + car2)
```

```python
#tuple unpacking
Person_details = ("vivek", 23, "chennai")
name,age,city = Person_details
print(name)
print(age)
print(city)
```

```python
#count elements
car = ("Tata", "Bugatti","Rolls Royce", "Mustang")
print(car.count("Tata"))
```

```python
#Nested tuple
car = (("Tata", "Bugatti"),("Rolls Royce", "Mustang"),("hyundai", "benz"))
print(car)
print(car[1])
```

```python
# Immutable
car = ("Tata", "Bugatti","Rolls Royce", "Mustang")
car[1] = "Benz"
print(car)
```

<h2>Set</h2>

A set is an unordered collection of unique items. It automatically removes duplicate values.

Unordered - Items do not have a fixed position. 
    
Unique - Duplicate values are not allowed

Imagine a classroom.

Five students write their favorite fruit.

Apple Mango Apple Orange Mango

If the teacher wants only the unique fruits, the result becomes:

Apple Mango Orange

This is exactly how a set works.

```python
# only unique items
numbers = {1,2,3,4,5,2}
print(numbers)
```

```python
#adding items
bikes = {"GT 650", "Kawasiki", "Hayabusa"}
bikes.add("Java")
print(bikes)
```

```python
#removing items
bikes = {"GT 650", "Kawasiki", "Hayabusa"}
bikes.remove("GT 650")
print(bikes)
```

```python
#union of 2 sets
bikes = {"GT 650", "Kawasiki", "Hayabusa",500}
cc = {500,1000, 1200}
print(bikes.union(cc))
```

```python
bikes = {"GT 650", "Kawasiki", "Hayabusa",500}
cc = {500,1000, 1200}
print(bikes.intersection(cc))
```

<h2>Dictionary</h2>

A dictionary is a collection that stores data as key-value pairs. Each key is unique and is used to access its corresponding value.

Example

Think of an English dictionary.
    
Word Meaning Apple A fruit Book A collection of pages

The word is the key.

The meaning is the value.

Similarly,

student = { "Name": "Nanda", "Age": 25 }

```python
student = {"Name":"Suresh", "Age": 23, "City": "Chennai"}
print(student)
```

```python
# Access vaues
student = {"Name":"Suresh", "Age": 23, "City": "Chennai"}
print(student["Name"])
print(student["Age"])
```

```python
#add a key-value pair
student = {"Name":"Suresh", "Age": 23}
student["City"] = "Chennai"
print(student)
```

```python
#remove an item
student = {"Name":"Suresh", "Age": 23, "City": "Chennai"}
student.pop("City")
```
